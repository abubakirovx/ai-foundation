```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("student_performance.csv")
print("Dataset shape (rows,columns) :",df.shape)
print("Column name :",df.columns.tolist())

print("First 5 rows of Dataset :")
print(df.head())

print("Summary statistics :")
print(df.describe())

print("Data types and null info:")
print(df.info())
```

```python
print("Avarage scores by columns")
df[["class_participation","total_score"]].mean()
```

```python
print("Maximum scores by columns")
df[["class_participation","total_score"]].max()
```

```python
print("Minimum scores by columns")
df[["class_participation","total_score"]].min()
```

```python
# Correlation
df[["class_participation","total_score"]].corr()
```

```python
df["total sum"]=df[["class_participation","total_score"]].sum(axis=1)
df["avarage score"]=df['total sum']/2
df["passed"]=np.where(df['avarage score']>=50 ,"Yes","No")

print(df[['total sum',"avarage score","passed"]].head())
```

```python
weekly_self_study_group=df.groupby("weekly_self_study_hours")[["class_participation","total_score"]].mean()
print(weekly_self_study_group)

attendance_percentage_group=df.groupby("attendance_percentage")[["class_participation","total_score"]].mean()
print(attendance_percentage_group)

pass_rate = df["passed"].value_counts(normalize=True)*100
print("pass_rate\n",pass_rate)
```

```python
# Convert scores to Numpy array
scores = df[["class_participation","total_score"]].values

# Mean ,Std Deviation
print("Numpy Mean",np.mean(scores,axis=0))
print("Numpy Std",np.std(scores,axis=0))

# Z-score (normalize)
z_score= scores-np.mean(scores,axis=0)/np.std(scores,axis=0)
print("Z-score (normalized numbers)\n",z_score)
#Avarage 0
#Negative -
#Positive +
```

```python
#How many students score above 90 avarage?
top_scores= df[df['avarage score']>=55]
print("Students score above 90 avarage,",len(top_scores))

#How many students failed exam?
failed= df[df['passed']=="No"]
print("Students failed exam,",len(failed))
```
